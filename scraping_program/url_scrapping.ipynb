```python
import requests
from bs4 import BeautifulSoup as bs
import pandas as pd
import csv
import re
from tqdm import tqdm
```

```python
df_page_names = pd.read_csv('page_names_transformed.csv')
```

```python
page_list = df_page_names['name'].to_list()
```

```python
test_list = page_list[:10]
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/87.0.4280.88 Safari/537.36'
}

session = requests.Session()

session.headers = headers
```

```python
url = f'https://www.immoweb.be/en/search/house-and-apartment/for-sale/{test_list[0]}?countries=BE&page=1&orderBy=relevance'

response = session.get(url)
soup = bs(response.text, "html.parser")

links = soup.find_all("a", class_="card__title-link")

print(len(links))

url_list = []

for link in links:
    url_list.append(link['href'])


print(url_list)
print(soup.prettify())
```

```python
def url_finder(url):
    response = session.get(url)
    soup = bs(response.text, "html.parser")

    h1 = soup.find("h1", string=re.compile("for sale"))
    parent = h1.find_parent("div").find_parent("div")
    links = parent.find_all("a", class_="card__title-link")   

    if len(links) == 0:
        return False
    
    else:
        url_list = []
        for link in links:
            url_list.append(link['href'])
        return url_list
```

```python
print(url_finder(f'https://www.immoweb.be/en/search/house-and-apartment/for-sale/{test_list[0]}?countries=BE&page=100&orderBy=relevance'))
```

```python
i = 1
result = url_finder(f'https://www.immoweb.be/en/search/house-and-apartment/for-sale/jette/1090?countries=BE&page={i}&orderBy=relevance')
while result is not False:
    with open('urls.csv', 'a', newline='') as file:
        for url in result:
            csv.writer(file).writerow([url])

    i+=1
    print(f"going to {i} page")
    result = url_finder(f'https://www.immoweb.be/en/search/house-and-apartment/for-sale/jette/1090?countries=BE&page={i}&orderBy=relevance')
```

```python
for page_name in tqdm(error_list):
    i = 1
    result = url_finder(f'https://www.immoweb.be/en/search/house-and-apartment/for-sale/{page_name}?countries=BE&page={i}&orderBy=relevance')
    while result is not False:
        with open('urls.csv', 'a', newline='') as file:
            for url in result:
                csv.writer(file).writerow([url])

        i+=1        
        result = url_finder(f'https://www.immoweb.be/en/search/house-and-apartment/for-sale/{page_name}?countries=BE&page={i}&orderBy=relevance')
```

```python
import pandas as pd
import re
```

```python
df_page_names = pd.read_csv('page_names.csv')
```

```python
df_page_names
```

```python
df_page_names = df_page_names[df_page_names['name'] != 'House and apartment for sale in Belgium - 10000 properties']
```

```python
df_page_names = df_page_names.map(lambda x: x.replace("House and apartment for sale - ", ""))
```

```python
df_page_names['name'][0].replace("House and apartment for sale - ", "")
```

```python
df_page_names['name'] = df_page_names.name.apply(func=lambda x: re.sub("\s-\s\d*\sproperties", "", str(x)))
```

```python
re.sub("\s-\s\d*\sproperties", "", df_page_names['name'][0])
```

```python
df_page_names['name'] = df_page_names['name'].apply(lambda x: x.replace(" (", "/"))
```

```python
df_page_names['name'] = df_page_names['name'].apply(lambda x: x.replace(")", ""))
```

```python
df_page_names['name'] = df_page_names['name'].apply(lambda x: x.replace(" ", "-"))
```

```python
df_page_names['name'] = df_page_names['name'].apply(lambda x: x.lower())
```

```python
df_page_names['name'] = df_page_names['name'].apply(lambda x: x.replace("---1-property", ""))
```

```python
df_page_names.to_csv('page_names_transformed.csv', index=False)
```
